```python
# ============================================================
# EMOTION DETECTION USING LSTM - COMPLETE PROJECT
# Dataset: DAIR.AI Emotion Dataset (Hugging Face)
# Emotions: Sadness, Joy, Love, Anger, Fear, Surprise
# ============================================================

# 1. INSTALL LIBRARIES
!pip install -q pyarrow wordcloud

# 2. IMPORT LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
import nltk

from google.colab import files
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from wordcloud import WordCloud

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Embedding, Dropout, Input
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

print("Libraries imported successfully!")


# 3. UPLOAD DATASET FILES
print("Upload train, test and validation Parquet files")

uploaded = files.upload()


# 4. LOAD DATASET
train_data = pd.read_parquet("train-00000-of-00001.parquet")
test_data = pd.read_parquet("test-00000-of-00001.parquet")
validation_data = pd.read_parquet("validation-00000-of-00001.parquet")

print("Train shape:", train_data.shape)
print("Test shape:", test_data.shape)
print("Validation shape:", validation_data.shape)


# 5. CONVERT EMOTION LABELS
emotion_labels = {
    0: "sadness",
    1: "joy",
    2: "love",
    3: "anger",
    4: "fear",
    5: "surprise"
}

for df in [train_data, test_data, validation_data]:
    df["Emotion"] = df["label"].map(emotion_labels)
    df.rename(columns={"text": "Comment"}, inplace=True)

print(train_data.head())


# 6. SAVE DATASET AS TXT FILES
train_data[["Comment", "Emotion"]].to_csv(
    "train.txt", sep=";", index=False, header=False
)

test_data[["Comment", "Emotion"]].to_csv(
    "test.txt", sep=";", index=False, header=False
)

validation_data[["Comment", "Emotion"]].to_csv(
    "val.txt", sep=";", index=False, header=False
)

print("TXT files created successfully!")


# 7. EXPLORE DATASET
print("\nEmotion Counts:")
print(train_data["Emotion"].value_counts())

print("\nMissing Values:")
print(train_data.isnull().sum())

print("\nDataset Information:")
print(train_data.info())


# 8. SENTENCE LENGTH DISTRIBUTION
train_data["length"] = train_data["Comment"].apply(len)

plt.figure(figsize=(10, 6))
sns.histplot(train_data["length"], kde=True)
plt.title("Sentence Length Distribution")
plt.xlabel("Sentence Length")
plt.ylabel("Frequency")
plt.show()


# 9. WORD CLOUD VISUALIZATION
def words_cloud(df, emotion):
    text = " ".join(
        df[df["Emotion"] == emotion]["Comment"].astype(str)
    )

    if not text.strip():
        return

    wc = WordCloud(
        width=800,
        height=600,
        background_color="white"
    ).generate(text)

    plt.figure(figsize=(7, 7))
    plt.imshow(wc)
    plt.axis("off")
    plt.title(f"{emotion} Word Cloud")
    plt.show()


for emotion in emotion_labels.values():
    words_cloud(train_data, emotion)


# 10. LABEL ENCODING
lb = LabelEncoder()

y_train = lb.fit_transform(train_data["Emotion"])
y_test = lb.transform(test_data["Emotion"])
y_val = lb.transform(validation_data["Emotion"])

print("Emotion Classes:", lb.classes_)


# 11. SEPARATE INPUT AND OUTPUT
X_train = train_data["Comment"]
X_test = test_data["Comment"]
X_val = validation_data["Comment"]


# 12. TEXT PREPROCESSING
vocabSize = 11000
maxLen = 300

def text_cleaning(df_column):
    stemmer = PorterStemmer()
    corpus = []

    for text in df_column:
        text = re.sub("[^a-zA-Z]", " ", str(text))
        text = text.lower().split()

        text = [
            stemmer.stem(word)
            for word in text
            if word not in stop_words
        ]

        text = " ".join(text)
        corpus.append(text)

    one_hot_words = [
        one_hot(word, vocabSize)
        for word in corpus
    ]

    padded = pad_sequences(
        one_hot_words,
        maxlen=maxLen,
        padding="pre"
    )

    return padded


# 13. CONVERT TEXT INTO SEQUENCES
x_train = text_cleaning(X_train)
x_test = text_cleaning(X_test)
x_val = text_cleaning(X_val)

print("Training Shape:", x_train.shape)
print("Testing Shape:", x_test.shape)
print("Validation Shape:", x_val.shape)


# 14. ONE HOT ENCODING
y_train = to_categorical(y_train, num_classes=6)
y_test = to_categorical(y_test, num_classes=6)
y_val = to_categorical(y_val, num_classes=6)


# 15. BUILD LSTM MODEL
model = Sequential()

model.add(Input(shape=(maxLen,)))
model.add(Embedding(vocabSize, 150))
model.add(Dropout(0.2))
model.add(LSTM(128))
model.add(Dropout(0.2))
model.add(Dense(64, activation="relu"))
model.add(Dropout(0.2))
model.add(Dense(6, activation="softmax"))

model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()


# 16. EARLY STOPPING
callback = EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)


# 17. TRAIN MODEL
history = model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=10,
    batch_size=64,
    callbacks=[callback],
    verbose=1
)


# 18. MODEL EVALUATION
train_loss, train_accuracy = model.evaluate(
    x_train, y_train, verbose=0
)

val_loss, val_accuracy = model.evaluate(
    x_val, y_val, verbose=0
)

test_loss, test_accuracy = model.evaluate(
    x_test, y_test, verbose=0
)

print("\nTraining Accuracy:", train_accuracy)
print("Validation Accuracy:", val_accuracy)
print("Testing Accuracy:", test_accuracy)


# 19. TRAINING ACCURACY GRAPH
acc = history.history["accuracy"]
val_acc = history.history["val_accuracy"]

epochs = range(1, len(acc) + 1)

plt.figure(figsize=(8, 5))
plt.plot(epochs, acc, label="Training Accuracy")
plt.plot(epochs, val_acc, label="Validation Accuracy")
plt.title("Training vs Validation Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.show()


# 20. TRAINING LOSS GRAPH
loss = history.history["loss"]
val_loss = history.history["val_loss"]

plt.figure(figsize=(8, 5))
plt.plot(epochs, loss, label="Training Loss")
plt.plot(epochs, val_loss, label="Validation Loss")
plt.title("Training vs Validation Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()


# 21. CONFUSION MATRIX
y_pred = model.predict(x_test, verbose=0)

y_pred_classes = np.argmax(y_pred, axis=1)
y_true = np.argmax(y_test, axis=1)

print("\nAccuracy Score:")
print(accuracy_score(y_true, y_pred_classes))

print("\nClassification Report:")
print(
    classification_report(
        y_true,
        y_pred_classes,
        target_names=lb.classes_,
        zero_division=0
    )
)

cm = confusion_matrix(y_true, y_pred_classes)

plt.figure(figsize=(8, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=lb.classes_,
    yticklabels=lb.classes_,
    cmap="Blues"
)

plt.xlabel("Predicted Emotion")
plt.ylabel("Actual Emotion")
plt.title("Emotion Detection Confusion Matrix")
plt.show()


# 22. CLEAN NEW SENTENCE
def clean_sentence(sentence):
    stemmer = PorterStemmer()

    text = re.sub("[^a-zA-Z]", " ", sentence)
    text = text.lower().split()

    text = [
        stemmer.stem(word)
        for word in text
        if word not in stop_words
    ]

    text = " ".join(text)

    seq = one_hot(text, vocabSize)

    padded = pad_sequences(
        [seq],
        maxlen=maxLen,
        padding="pre"
    )

    return padded


# 23. PREDICT EMOTION
def predict_emotion(sentence):
    processed = clean_sentence(sentence)

    pred = model.predict(processed, verbose=0)

    emotion = lb.classes_[np.argmax(pred)]

    confidence = np.max(pred) * 100

    print("\nInput Sentence:", sentence)
    print("Predicted Emotion:", emotion)
    print("Confidence:", round(confidence, 2), "%")


# 24. TEST NEW SENTENCES
predict_emotion("I am very happy today!")
predict_emotion("I am feeling very sad.")
predict_emotion("I am angry with my friend.")
predict_emotion("I am afraid of the dark.")
predict_emotion("I love spending time with my family.")
predict_emotion("Wow, this is an amazing surprise!")


# 25. SAVE TRAINED MODEL
model.save("emotion_detection_lstm.keras")

print("\nModel saved successfully!")


# 26. DOWNLOAD TRAINED MODEL
files.download("emotion_detection_lstm.keras")

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
